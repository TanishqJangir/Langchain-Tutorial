```python
from langchain_chroma import Chroma
from langchain_openai import OpenAIEmbeddings
from langchain_core.documents import Document
```

```python
documents = [
    Document(page_content="LangChain helps developers to build LLM applications easily."),
    Document(page_content="Chroma is a vector database opimized for LLM-based search."),
    Document(page_content="Embeddings convert text into high-dimensional vectors."),
    Document(page_content="OpenAI privides powerful embedding models."),
]
```

```python
embedding_model = OpenAIEmbeddings()

vectorstore = Chroma.from_documents(
    documents = documents,
    embedding = embedding_model,
    collection_name = "my_collection"
)
```

```python
retriever = vectorstore.as_retriever(search_kwargs={"k": 2})
```

```python
query = "what is chroma used for?"
result = retriever.invoke(query)
```

```python
for i , doc in enumerate(result):
    print(f"\n --Result {i+1} --")
    print(f"Content: \n{doc.page_content}...") #truncate for display
```

```python
otherres = vectorstore.similarity_search(query, k=2)

for i, doc in enumerate(otherres):
    print(f"\n --Result {i+1} --")
    print(f"Content: \n{doc.page_content}...") #truncate for display



#yes vector store has the capability for similarity search, but it has only one method for retrieval. The retriever search method is a more general method that can be used for other purposes as well.
```
