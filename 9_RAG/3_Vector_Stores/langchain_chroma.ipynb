```python
from langchain_openai import OpenAIEmbeddings
from langchain_chroma import Chromam
```

```python
from langchain_classic.schema import Document


doc1 = Document(page_content="Virat Kohli is one of the most successful and consistent batsmen in IPL history. Known for his aggressive batting style and exceptional performance under pressure.", metadata={"team": "Royal Challengers Bangalore"})

doc2 = Document(page_content="Rohit Sharma is the most successful captain in IPL history, leading Mumbai Indians to five titles. He is known for his leadership skills and match-winning performances.", metadata={"team": "Mumbai Indians"})

doc3 = Document(page_content="MS Dhoni is a legendary cricketer and former captain of the Indian national team. He is known for his calm demeanor, sharp cricketing acumen, and finishing abilities in high-pressure situations.", metadata={"team": "Chennai Super Kings"})

doc4= Document(page_content="Jasprit Bumrah is one of the best fast bowlers in the world, known for his unique bowling action and ability to bowl yorkers consistently. He has been a key player for the Indian cricket team in all formats.", metadata={"team": "Mumbai Indians"})

doc5 = Document(page_content="Ravindra Jadeja is an all-rounder known for his exceptional fielding, left-arm spin bowling, and ability to contribute with the bat. He has been a crucial player for the Indian cricket team and Chennai Super Kings.", metadata={"team": "Chennai Super Kings"})
```

```python
docs = [doc1, doc2, doc3, doc4, doc5]
```

```python
vector_store = Chroma(
    embedding_function=OpenAIEmbeddings(),
    persist_directory="chroma_db",
    collection_name="cricket_players"
)
```

```python
vector_store.add_documents(docs)
```

```python
vector_store.get(include=["embeddings", "metadatas", "documents"])
```

```python
vector_store.similarity_search(
    query="Who among these are a bowler?",
    k=2
)
```

```python
#search with similarity score

vector_store.similarity_search_with_score(
    query="Who among these are a bowler?",
    k=2
)
```

```python
#meta-data filtering

vector_store.similarity_search_with_score(
    query="",
    filter={"team": "Mumbai Indians"},
)
```

```python
updated_doc1 = Document(
    page_content="Virat Kohli is one of the most successful and consistent batsmen in IPL historyyyyyy. Known for his aggressive batting style and exceptional performance under pressure.",
    metadata={"team": "Royal Challengers Bangalore", "role": "batsman"}
)

vector_store.update_document(document_id="5fb08761-53c9-4e23-8e97-81cd304f7cc5", document=updated_doc1)
```
