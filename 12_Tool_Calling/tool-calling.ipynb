```python
from langchain_openai import ChatOpenAI
from langchain_core.tools import tool
from langchain_core.messages import HumanMessage
import requests
from pprint import pprint
```

```python
# create a tool

@tool
def multiply(a: int, b: int) -> int:
    """Given two integers a and b, this tool returns the product of a and b."""
    return a * b
```

```python
print(multiply.invoke({"a": 3, "b": 4}))  # Output: 12
```

```python
print(multiply.name)
print(multiply.description)
print(multiply.args)
```

```python
#tool binding

llm = ChatOpenAI()
```

```python
llm_with_tools = llm.bind_tools([multiply])
```

```python
res = llm_with_tools.invoke("Hii how are you")
pprint(res.model_dump())
```

```python
query = HumanMessage("Can you multiply 3 with 10000")
```

```python
messages = [query]
```

```python
messages
```

```python
response = llm_with_tools.invoke(messages)
```

```python
messages.append(response)
```

```python
messages
```

Tool Execution

```python
tool_result = multiply.invoke(response.tool_calls[0])
```

```python
messages.append(tool_result)
```
