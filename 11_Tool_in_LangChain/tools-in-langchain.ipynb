Custom Tools

```python
from langchain_core.tools import tool 
```

```python
#Step 1 - Create a function

def multiply(a, b):
    return a * b
```

```python
#Step 2 - Add type hints to the function

def multiply(a: int, b: int) -> int:
    return a * b
```

```python
#Step 3 - Add tool decorator to the function

# Docstring - A docstring is a string literal that occurs as the first statement in a module, function, class, or method definition. Such a docstring becomes the __doc__ special attribute of that object. The docstring should describe the function's behavior and its parameters.

@tool
def multiply(a: int, b: int) -> int:
    """Multiply two integers."""  #its mandatory to add the docstring to the function
    return a * b
```

```python
result = multiply.invoke({"a": 5, "b": 3})
```

```python
print(result)
```

```python
print(multiply.name)
print(multiply.description)
print(multiply.args)
```

```python
print(multiply.args_schema.model_json_schema())
```

Method 2 - Using Structured Tool

```python
from langchain_core.tools import StructuredTool
from pydantic import BaseModel, Field
```

```python
class MultiplyInput(BaseModel):
    a: int = Field(required=True, description="The first number to multiply.")
    b: int = Field(required=True, description="The second number to multiply.")
```

```python
def multiply_func(a: int, b: int) -> int:
    return a * b
```

```python
multiply_tool = StructuredTool.from_function(
    func=multiply_func,
    name="multiply",
    description="Multiply two numbers.",
    args_schema=MultiplyInput
)
```

```python
result = multiply_tool.invoke({"a": 5, "b": 3})

print(result)
print(multiply_tool.name)
print(multiply_tool.description)
print(multiply_tool.args)
```

Method 3 - Using Base Tool class

```python
from langchain_core.tools import BaseTool
from typing import Type
```

```python
# args schema using pydantic

class MultiplyInput(BaseModel):
    a: int = Field(required=True, description="The first number to multiply.")
    b: int = Field(required=True, description="The second number to multiply.")
```

```python
class MultiplyTool(BaseTool):
    name : str = "multiply"
    description : str = "Multiply two numbers."
    args_schema : Type[BaseModel] = MultiplyInput

    def _run(self, a: int, b:int) -> int:
        return a * b
```

```python
multiply_tool = MultiplyTool()
```

```python
result = multiply_tool.invoke({"a": 5, "b": 3})
```

```python
print(result)
print(multiply_tool.name)
print(multiply_tool.description)
```

Toolkits

```python
from langchain_core.tools import tool
```

```python
# custom tools

@tool
def add(a: int, b: int) -> int:
    """Add two integers."""
    return a + b

@tool
def multiply(a: int, b: int) -> int:
    """Multiply two integers."""
    return a * b
```

```python
class MathToolkit:
    def get_tools(self):
        return [add, multiply]
```

```python
toolkit = MathToolkit()
tools = toolkit.get_tools()

for tool in tools:
    print(f"Tool Name: {tool.name}")
    print(f"Tool Description: {tool.description}")
```
