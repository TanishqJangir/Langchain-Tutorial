```python
from dotenv import load_dotenv
import os
load_dotenv()

openai_api_key = os.getenv("OPENAI_API_KEY")
```

```python
from youtube_transcript_api import YouTubeTranscriptApi
from youtube_transcript_api._errors import TranscriptsDisabled
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_openai import OpenAIEmbeddings, ChatOpenAI
from langchain_community.vectorstores import FAISS
from langchain_core.prompts import PromptTemplate
```

```python
video_id = "lVylRtlPOIE"  # Only the id, not full url
try:

    yt_api = YouTubeTranscriptApi()
    transcript_list = yt_api.fetch(
        video_id, 
        languages=['en']
    )

    transcript = " ".join(chunk.text for chunk in transcript_list)

    print(transcript)

except TranscriptsDisabled:
    print("No captions available for this video.")
```

```python
transcript_list
```

```python
splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000, 
    chunk_overlap=200
)
chunks = splitter.create_documents([transcript])
```

```python
len(chunks)
```

```python
print(chunks[10].page_content)
```

```python
embeddings = OpenAIEmbeddings(model="text-embedding-3-large")
vectorstore = FAISS.from_documents(chunks, embeddings)
```

```python
vectorstore.index_to_docstore_id
```

```python
vectorstore.get_by_ids(['0bb8caca-657b-4916-8b4f-35a8a8bfdfe4'])
```

```python
retriever = vectorstore.as_retriever(search_type="similarity", search_kwargs={"k": 4})
```

```python
retriever
```

```python
result = retriever.invoke("What are the kids doing in the video?")
```

```python
print(result[0].page_content)
```

Step 3 - Augmentation

```python
llm = ChatOpenAI(
    model="gpt-4o-mini",
    temperature=0.2,
)
```

```python
prompt = PromptTemplate(
    template="""
        You are a helpful assistant.
        Answer Only from the provided transcript context.
        If the context does not contain the answer, just say "I don't know".

        {context}
        Question: {question}
    """,

    input_variables=["context", "question"]
)
```

```python
question = "Is the topic of kids discussed in this video? If yes then what was discussed"
retrieved_docs = retriever.invoke(question)
```

```python
context_text = "\n \n".join([doc.page_content for doc in retrieved_docs])
context_text
```

```python
final_prompt = prompt.invoke({"context" : context_text, "question": question})
```

Step 4 - Generation

```python
answer = llm.invoke(final_prompt)
print(answer.content)
```

Building a Chain

```python
from langchain_core.runnables import RunnableParallel, RunnablePassthrough, RunnableLambda
from langchain_core.output_parsers import StrOutputParser
```

```python
def format_docs(retrieved_docs):
    context_text = "\n \n".join([doc.page_content for doc in retrieved_docs])
    return context_text
```

```python
parallel_chain = RunnableParallel({
    "context" : retriever | RunnableLambda(format_docs),
    "question" : RunnablePassthrough()
})
```

```python
vidresult = parallel_chain.invoke("Whose video is this? and what is it about?")
```

```python
print(vidresult['context'])
```

```python
parser = StrOutputParser()
```

```python
main_chain = parallel_chain | prompt | llm | parser
```

```python
main_chain.invoke("Can you summarize the video and tell me what is it about?")
```
